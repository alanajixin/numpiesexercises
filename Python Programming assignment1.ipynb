````python
#Addition of two numbers
def addition(num1,num2):
    return num1+num2

#subtraction of 2 numbers
def subtraction(num1,num2):
    return num1-num2

#multiplication
def multiplication(num1,num2):
    return num1*num2

#division
def division(num1,num2):
    if num2==0:
        print("Number cannot be divided by zero")
    else:
        return num1/num2


#Read input number_1 and number_2
while True:
    number_1=int(input("Enter the first Number"))
    number_2=int(input("Enter the second Number"))
    operation=input("which operation you need (+, -, *, /)")
    if operation== '+':
        print(addition(number_1,number_2))
    elif operation== '-':
        print(subtraction(number_1,number_2))
    elif operation== '*':
        print(multiplication(number_1,number_2))
    else:
        print(division(number_1,number_2))
    cont=input("Do You want to continue(y/n)")
    if cont=='n':
        break
        
````

````python
#create a list for name and mark
names=['Peter','Mark','John','Angel','Reshma']
marks=[84,74,86,94,70]
students_performance=dict(zip(names,marks))
print(students_performance)

# Display all student names
print("All Student names are")
for x in students_performance.keys():
    print(x,end=", ")
    
#Display all marks
print("\n Different  Marks obtained by students are")
for x in students_performance.values():
    print(x,end=",")
print("\n")

#Student with highest mark

highest_mark=max(students_performance.values())

for name, mark in students_performance.items() :
    if mark == highest_mark:
        student_highest_mark=name
print(f"Student Obtained highest Mark is {student_highest_mark} having {highest_mark} ")
````

````python
sentence=input("Enter any sentence")
 # convert sentence to uppercase
print("The sentence converted to upper case is : ", sentence.upper())

#count number of vowels in sentence
vowels=['a','e','i','o','u']
count_vowels=0
for letter in sentence.lower():
    if letter in vowels:
        count_vowels+=1
print("Number of vowels are :", count_vowels)

#reverse sentence 
reversed_words = " ".join(sentence.split()[::-1])
print("words in reverse form:", reversed_words)
print("Sentence in reverse form:",sentence[::-1])
````

````python
"""Write a program that:
- Creates a Pandas DataFrame with a column of dates in the format `YYYY-MM-DD`.
- Extracts the year, month, and day from each date.
- Calculates the difference between two dates."""

df = pd.DataFrame({
    'Date': ['2026-01-15', '2026-03-20', '2026-04-10', '2026-06-19']
})
print(df.info())
#Here we can see time is here of dtype object
#so we change to datetime
df['Date']=pd.to_datetime(df['Date'])

print(df.info())

#Extracts the year, month, and day from each date.
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day
print("DataFrame after extracting Year, Month, and Day:")
print(df)
#Calculates the difference between two dates
difference=df['Date'][1]-df['Date'][0]
print("Difference between dates:")
print(difference)
print("Number of days:", difference.days)
````

````python
"""Handling Missing Data:
Create a DataFrame with some missing values. Write a program to:
- Identify the missing values.
- Fill the missing values with the mean of the respective column.
- Drop any rows that still contain missing values."""
data={'Name':['John','Peter','Anna','Luke','Kevin'],'Age': [25, np.nan, 30, 35, np.nan],
    'Salary': [50000, 60000, np.nan, 80000, 70000],
    'Experience': [2, 4, np.nan, 8, 5]}
data_employee=pd.DataFrame(data)

#Identify the missing values.
print("Missing values are")
print(data_employee.info())
#Total Missing values
print("Total missing values")
print(data_employee.isnull().sum())

# Fill missing values with the mean of each column
data_employee['Salary'] = data_employee['Salary'] .fillna(data_employee['Salary'].mean())
data_employee['Age']= data_employee['Age'] .fillna(data_employee['Age'].mean())
print("\nDataFrame after filling missing values with mean:")
print(data_employee)

# Drop rows that still contain missing values
df_cleaned = data_employee.dropna()

print("\nDataFrame after dropping remaining missing values:")
print(df_cleaned)
````
